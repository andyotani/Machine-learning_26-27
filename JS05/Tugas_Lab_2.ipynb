```python
# Persiapan Dataset
# Import Library
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split

from sklearn.feature_extraction.text import CountVectorizer
from sklearn.feature_extraction.text import TfidfVectorizer

from sklearn.naive_bayes import MultinomialNB

from sklearn.metrics import accuracy_score
from sklearn.metrics import confusion_matrix
from sklearn.metrics import classification_report
```

```python
# Load Dataset
data = pd.read_csv(
    "spam.csv",
    encoding="latin-1"
)

data.head()
```

Dataset memiliki beberapa kolom yang tidak digunakan, sehingga hanya digunakan:

* v1 → label spam/ham
* v2 → isi SMS

```python
# Menghapus kolom tidak diperlukan
data = data[['v1','v2']]

data.columns = [
    'label',
    'SMS'
]

data.head()
```

```python
# Encoding Label

# Label diubah menjadi numerik:
# ham = 0
# spam = 1

data['label'] = data['label'].map(
    {
        'ham':0,
        'spam':1
    }
)
```

```python
# Memisahkan fitur dan label
X = data['SMS']

y = data['label']
```

```python
# Membagi Data Training dan Testing
# Data dibagi: 80% training dan 20% testing
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.2,
    random_state=50
)
```

### Percobaan 1 : Multinomial Naive Bayes + CountVectorizer

```python
# Ekstraksi fitur CountVectorizer
# CountVectorizer mengubah teks SMS menjadi representasi jumlah kemunculan kata.
# Stop words diaktifkan untuk menghilangkan kata yang kurang memberikan informasi.
count_vectorizer = CountVectorizer(
    stop_words='english'
)


X_train_count = count_vectorizer.fit_transform(
    X_train
)


X_test_count = count_vectorizer.transform(
    X_test
)
```

```python
# Membuat Model Multinomial Naive Bayes
model_count = MultinomialNB()


model_count.fit(
    X_train_count,
    y_train
)
```

```python
# Evaluasi Model
y_pred_count = model_count.predict(
    X_test_count
)


print(
    "Accuracy:",
    accuracy_score(
        y_test,
        y_pred_count
    )
)


print(
    confusion_matrix(
        y_test,
        y_pred_count
    )
)


print(
    classification_report(
        y_test,
        y_pred_count
    )
)
```

Hasil CountVectorizer

Accuracy: 98.30%
Confusion Matrix: [[951   3]
 [ 16 145]]

 Interpretasi:

Terdapat 19 data yang mengalami kesalahan klasifikasi, yaitu 3 data ham diprediksi sebagai spam dan 16 data spam diprediksi sebagai ham.

### Percobaan 2 : Multinomial Naive Bayes + TF-IDF

```python
# 1. Ekstraksi Fitur TF-IDF
# TF-IDF mempertimbangkan bobot pentingnya suatu kata berdasarkan frekuensi kemunculan kata tersebut.
tfidf_vectorizer = TfidfVectorizer(
    stop_words='english'
)


X_train_tfidf = tfidf_vectorizer.fit_transform(
    X_train
)


X_test_tfidf = tfidf_vectorizer.transform(
    X_test
)
```

```python
# Membuat Model Naive Bayes
model_tfidf = MultinomialNB()


model_tfidf.fit(
    X_train_tfidf,
    y_train
)
```

```python
# Evaluasi Model
y_pred_tfidf = model_tfidf.predict(
    X_test_tfidf
)


print(
    "Accuracy:",
    accuracy_score(
        y_test,
        y_pred_tfidf
    )
)


print(
    confusion_matrix(
        y_test,
        y_pred_tfidf
    )
)


print(
    classification_report(
        y_test,
        y_pred_tfidf
    )
)
```

Hasil TF-IDF

Accuracy: 96.05%
Confusion Matrix: [[954   0]
 [44 117]]

 Interpretasi:

Seluruh data ham berhasil diklasifikasikan dengan benar, tetapi masih terdapat 44 data spam yang salah diklasifikasikan sebagai ham.

### Perbandingan Hasil Model

| Metode Fitur | Algoritma | Accuracy |
|---|---|---|
| CountVectorizer + stop_words | Multinomial Naive Bayes | **98.30%** |
| TF-IDF + stop_words | Multinomial Naive Bayes | 96.05% |





### Kesimpulan
Berdasarkan hasil pengujian, model Multinomial Naive Bayes dengan fitur CountVectorizer dan stop_words menghasilkan performa terbaik dengan akurasi sebesar 98.30%, sedangkan penggunaan fitur TF-IDF menghasilkan akurasi sebesar 96.05%. Pada kasus klasifikasi SMS spam, CountVectorizer lebih optimal karena representasi berdasarkan frekuensi kemunculan kata mampu menangkap pola kata yang sering muncul pada pesan spam. Oleh karena itu, fitur terbaik untuk dataset spam.csv adalah CountVectorizer dengan stop_words menggunakan algoritma Multinomial Naive Bayes.

Dengan demikian, kombinasi terbaik untuk klasifikasi spam SMS pada dataset ini adalah CountVectorizer dengan stop_words menggunakan algoritma Multinomial Naive Bayes dengan akurasi sebesar 98.30%.
